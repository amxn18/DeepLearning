```python
import numpy as np
import torch
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score
```

### Scaling

```python
scaler = StandardScaler()
xTrain = scaler.fit_transform(xTrain)
xTest = scaler.transform(xTest)
```

```python
encoder = LabelEncoder()
yTrain = encoder.fit_transform(yTrain)
yTest = encoder.transform(yTest)
```

* Converting into tensors

```python
xTrainT = torch.from_numpy(xTrain.astype(np.float32))
xTestT = torch.from_numpy(xTest.astype(np.float32))
yTrainT = torch.from_numpy(yTrain.astype(np.float32))
yTestT = torch.from_numpy(yTest.astype(np.float32))
```

```python
xTrainT.shape
```

# Defining the model

```python
class NNmodel():
  def __init__(self, X):
    self.weights = torch.rand(x.shape[1], 1, dtype = torch.float32, requires_grad= True) # 30x1
    self.bias = torch.zeros(1, dtype = torch.float32, requires_grad= True)

  def forward(self, X):
    z = torch.matmul(X, self.weights) + self.bias
    yPred = torch.sigmoid(z)
    return yPred

  def lossFunc(self, yPred, yTrue):
    eps = 1e-7
    yPred = torch.clamp(yPred, eps, 1 - eps)
    return -(yTrue * torch.log(yPred) + (1 - yTrue) * torch.log(1 - yPred)).mean()

```

```python
learningRate = 0.1
epochs = 30
```

# Training Pipeline

```python
model = NNmodel(xTrainT)
print(model.weights)
print(model.weights.shape)
print(model.bias)
print(model.bias.shape)
```

# Training Model

```python
for epoch in range(epochs):

    # Forward Pass
    yPred = model.forward(xTrainT)

    # Calculate Loss
    loss = model.lossFunc(yPred, yTrainT)

    # Backpropogation
    loss.backward()

    # Update weights
    with torch.no_grad():
        model.weights.data -= learningRate * model.weights.grad
        model.bias.data    -= learningRate * model.bias.grad

    # Reset gradients
    model.weights.grad.zero_()
    model.bias.grad.zero_()

    print(f"Epoch {epoch+1}, Loss: {loss.item():.4f}")
```

```python
# model evaluation
with torch.no_grad():
  yPred = model.forward(xTestT)
  yPred = (yPred > 0.9).float()
  accuracy = (yPred == yTestT).float().mean()
  print(f'Accuracy: {accuracy.item()}')
```
